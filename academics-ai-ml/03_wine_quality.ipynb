# Red Wine Quality Prediction

**Goal:** Predict wine quality score + Good/Bad classification
**Algorithm:** Random Forest
**Dataset:** [Red Wine Quality](https://www.kaggle.com/datasets/uciml/red-wine-quality-cortez-et-al-2009)

```python
import kagglehub
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, accuracy_score
```

## 1. Load Data

```python
path = kagglehub.dataset_download('uciml/red-wine-quality-cortez-et-al-2009')
df = pd.read_csv(f'{path}/winequality-red.csv', sep=';')
print('Shape:', df.shape)
print('Quality:', df['quality'].value_counts().sort_index().to_dict())
```

## 2. Split

```python
X = df.drop('quality', axis=1)
y_reg = df['quality']
y_clf = (df['quality'] >= 7).astype(int)
X_train, X_test, yr_train, yr_test = train_test_split(X, y_reg, test_size=0.2, random_state=42)
_, _, yc_train, yc_test = train_test_split(X, y_clf, test_size=0.2, random_state=42)
```

## 3. Train

```python
reg = RandomForestRegressor(n_estimators=200, random_state=42)
reg.fit(X_train, yr_train)
print('RMSE: %.3f' % np.sqrt(mean_squared_error(yr_test, reg.predict(X_test))))
```

```python
clf = RandomForestClassifier(n_estimators=200, random_state=42)
clf.fit(X_train, yc_train)
print('Accuracy: %.4f' % accuracy_score(yc_test, clf.predict(X_test)))
```
